Breast Cancer Classification with Neural Networks

Importing the dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

Data Collection & Processing

```python
breast_cancer_data = sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_data)
```

Loading the data into a dataframe

```python
df = pd.DataFrame(breast_cancer_data.data , columns = breast_cancer_data.feature_names)
```

```python
df.head()
```

```python
df['label'] = breast_cancer_data.target
```

```python
df.tail()
```

```python
df.shape
```

```python
df.info()
```

Checking for Null values

```python
df.isnull().sum()
```

Checking the Statistical Measures

```python
df.describe()
```

Checking the distribution of target variable

```python
df['label'].value_counts()
```

Description of Target Variable

1 --> Benign   
0 --> Malignant

```python
df.groupby('label').mean()
```

Separating the features and targets

```python
x = df.drop(columns="label" ,axis=1)

y = df["label"]
```

```python
print(x)
```

```python
print(y)
```

Splitting data into Training data and Test data

```python
x_train, x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state =2)
```

```python
print(x_train.shape , x_test.shape , x.shape)
```

Standardization of Data

```python
scaler = StandardScaler()

x_train_standard = scaler.fit_transform(x_train)

x_test_standard = scaler.transform(x_test)
```

Building the Neural Network

Importing Tensorflow and Keras

```python
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

Setting the layers of the Neural Network

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)),
    keras.layers.Dense(20, activation="relu"),
    keras.layers.Dense(2, activation="sigmoid")
])
```

Compiling the Neural Network

```python
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])
```

Training the Neural Network

```python
history = model.fit(x_train_standard , y_train , validation_split=0.1 , epochs=10)
```

Visualizing Accuracy and Loss

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Training Data','Validation Data'], loc='lower right')
plt.show()
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Training Data','Validation Data'], loc='upper right')
plt.show()
```

Accuracy of the Model on Test Data

```python
loss , accuracy = model.evaluate(x_test_standard , y_test)

print("Accuracy :" , accuracy)
print("Loss :" , loss)
```

```python
y_pred = model.predict(x_test_standard)
```

Converting the prediction probability to class labels

```python
y_pred_labels = [ np.argmax(i) for i in y_pred ]

print(y_pred_labels)
```

Building a predictive system

```python
input_data = (13.54,14.36,87.46,566.3,0.09779,0.08129,0.06664,0.04781,0.1885,0.05766,0.2699,0.7886,2.058,23.56,0.008462,0.0146,0.02387,0.01315,0.0198,0.0023,15.11,19.26,99.7,711.2,0.144,0.1773,0.239,0.1288,0.2977,0.07259)

input_data_as_np_array = np.asarray(input_data)

input_data_reshaped = input_data_as_np_array.reshape(1,-1)

input_data_std = scaler.transform(input_data_reshaped)

prediction = model.predict(input_data_std)

print(prediction)

prediction_label = [np.argmax(prediction)]

print(prediction_label)

if (prediction_label[0] == 0):
    print("The Tumor is Malignant")
else:
    print("The Tumor is Benign")
```
